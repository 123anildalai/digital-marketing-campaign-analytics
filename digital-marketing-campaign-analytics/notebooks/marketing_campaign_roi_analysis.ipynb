### Project Overview
Analyzing multi-channel advertising performance across 4,000+ campaign runs to optimize customer acquisition costs (CAC) and maximize Return on Ad Spend (ROAS).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/cleaned_marketing_data.csv")
print(f"Campaign Records: {df.shape[0]}")
df.head()
```

## 1. Full-Funnel Marketing Performance Summary

```python
total_spend = df["ad_spend"].sum()
total_revenue = df["revenue"].sum()
blended_roas = total_revenue / total_spend
blended_ctr = (df["clicks"].sum() / df["impressions"].sum()) * 100
avg_cpc = total_spend / df["clicks"].sum()

print(f"Total Media Investment: ₹{total_spend:,.2f}")
print(f"Total Revenue Output   : ₹{total_revenue:,.2f}")
print(f"Blended Portfolio ROAS : {blended_roas:.2f}x")
print(f"Overall CTR            : {blended_ctr:.2f}%")
print(f"Average CPC            : ₹{avg_cpc:.2f}")
```

## 2. Channel Profitability & Efficiency (ROAS Ranking)

```python
ch_breakdown = df.groupby("channel").agg(
    spend=("ad_spend", "sum"),
    revenue=("revenue", "sum"),
    conversions=("conversions", "sum")
).reset_index()
ch_breakdown["roas"] = (ch_breakdown["revenue"] / ch_breakdown["spend"]).round(2)
ch_breakdown["cac"] = (ch_breakdown["spend"] / ch_breakdown["conversions"]).round(2)
ch_breakdown.sort_values(by="roas", ascending=False)
```
